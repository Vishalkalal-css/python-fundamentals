#Operators in python


```python
#arthemetic Operator
print(5+4)

print(5-4)

print(5*4)

print(5/4)

print(5//4)

print(5%4)

print(5**4)
```

```python
#relational operator
print(4>5)

print(4<5)

print(4<=5)

print(4>=5)

print(4==5)

print(4!=5)
```

```python
#logical operator
print(1 and 0)

print(1 or 0)

print( not 1)
```

```python
#Bitwise operator

#Bitwise and
print(2 & 3)

#bitwise or
print(2 | 3)

#bitwise xor
print(2 ^ 3)

#bitwise not
print(~3)

#bitwise left shift
print(4 >> 2)

#bitwise right shift
print(5 << 2)
```

```python
# Gussing game

# generate a random integer between 1 and 100

import random
jakepot = random.randint(1,100)

guess = int(input('enter your guess '))
counter = 1
while guess != jakepot:
  if guess < jakepot:
    print('guess higher')
  else:
    print('guess lower')
  guess = int(input('enter your guess '))
  counter += 1
else:
  print('correct guess')
  print('attempts',counter)
```
